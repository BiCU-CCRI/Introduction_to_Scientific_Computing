## 1. Import packages

```python
%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Packages imported successfully.")
```

## 2. Create some example data

This code creates a small dataset containing study hours and exam scores.

```python
data = pd.DataFrame({
    "student": ["Alex", "Blair", "Casey", "Drew", "Emery", "Frankie"],
    "study_hours": [1, 2, 3, 4, 5, 6],
    "exam_score": [52, 58, 65, 73, 81, 88]
})

data
```

## 3. Explore the data

Use common pandas methods to inspect and summarise the dataset.

```python
# Show the first few rows
data.head()
```

```python
# Show summary statistics
data.describe()
```

## 4. Filter and add a new column

Here we select students who studied for at least four hours and create a pass/fail column.

```python
students_with_more_study = data[data["study_hours"] >= 4]
students_with_more_study
```

```python
data["passed"] = data["exam_score"] >= 60
data
```

## 5. Make a simple plot

This scatter plot shows the relationship between study hours and exam score.

```python
plt.figure(figsize=(7, 4))
plt.scatter(data["study_hours"], data["exam_score"])
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.title("Study Hours vs Exam Score")
plt.grid(True)
plt.show()
plt.close()
```

## 6. Try loading the plotly package

Now we will try to load a package that isn't included in the default environment. What happens?  

```python
import plotly.express as px
```

```python
# Make the same graph but with plotly

fig = px.scatter(
    data,
    x="study_hours",
    y="exam_score",
    hover_name="student",
    title="Study Hours vs Exam Score",
    labels={
        "study_hours": "Study hours",
        "exam_score": "Exam score"
    }
)

fig.show()
```
